```python
def img_to_channels(image):

    # convert RGB to HSV
    hsv_img = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

    # split the channels
    h, s, v = cv2.split(hsv_img)
    b, g, r = cv2.split(image)

    channel_dict = {
        "h": h,
        "s": s,
        "v": v,
        "b": b,
        "g": g,
        "r": r
    }

    return channel_dict

def ELMHighCoverage(image):
    channel_dict = img_to_channels(image)

    plt.scatter(channel_dict.get("h"), channel_dict.get("s"), channel_dict.get("v"))

    plt.show()



```
